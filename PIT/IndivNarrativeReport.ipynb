# Deep Learning Approach for Automated Identification of Blood Cell Types Using CNN-Based Multi-Class Classification

**Authors: John Vincent Gamali, Jezzel Faith Q. Gier, Keith Laspoña, and Kimberly Vocales**
 **Department of Data Science, USTP – Cagayan de Oro**
---

Narrative Report on “Deep Learning Approach for Automated Identification of Blood
Cell Types Using CNN-Based Multi-Class Classification”


The development of the research project titled Deep Learning Approach for Automated
Identification of Blood Cell Types Using CNN-Based Multi-Class Classification began with a
series of initial discussions aimed at establishing a shared understanding of its scope, objectives,
and scientific relevance. The group examined the need for automated white blood cell
identification, noting the limitations of manual blood smear examination and the increasing
demand for faster, more reliable diagnostic support in clinical settings. To deepen their
understanding, each member reviewed major literature on convolutional neural networks and
medical image analysis, including key works by LeCun et al. (2015) and Litjens et al. (2017).
These early efforts allowed the team to refine the research problem, strengthen the background
of the study, and clearly articulate its purpose and objectives.


As the literature review progressed, the team began drafting the introductory components
of the research paper. These sections explained the context of blood cell classification, provided
background information on white blood cell types, and highlighted the importance of automated
diagnostic systems in modern healthcare. The discussion also addressed the limitations of
manual microscopy and introduced the potential of CNN-based methods in improving diagnostic
accuracy. The significance of the study was emphasized, particularly its benefits to healthcare
practitioners, diagnostic laboratories, and the wider field of artificial intelligence in medicine.
One challenge encountered during this stage involved maintaining consistency in academic
writing style, as different members were assigned to different sections of the draft. The team
resolved this by collectively reviewing and editing each other’s submissions, ensuring uniformity
in tone, clarity, and formatting.


Collaboration played a central role in the group’s progression. Responsibilities were
informally distributed based on individual strengths, and regular meetings were conducted to
discuss concepts, clarify technical concerns, and evaluate draft outputs. These sessions allowed
the group to refine ideas collectively and address challenges more efficiently, strengthening both
the quality of the content and the teamwork dynamic.
In conclusion, the group has made considerable progress in establishing the foundational
components of the study. While the project remains in the research and drafting phase, the team
has successfully built a strong theoretical framework that will guide the next stages of the work,
including dataset preparation, preprocessing, and model development. Through continuous
collaboration, open communication, and shared commitment to academic rigor, the group is
steadily shaping a well-structured and coherent research project.
